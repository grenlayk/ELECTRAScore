```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset("social_i_qa")
```

```python
answer_labels = ['answerA', 'answerB', 'answerC']
socialiqa = dataset['train'].map(
    lambda row : {'answer': row[answer_labels[int(row['label']) - 1]]},
    remove_columns=["answerA", "answerB", "answerC", "label"]
)
socialiqa
```

```python
# define formatting functions
def format_row(row):
    question = f"question : {row['question']}"
    answer = f"answer : {row['answer']}"
    context = f"context : {row['context']}"
    text = f"{question}\n{answer}\n{context}\n"
    return text
```

```python
socialiqa_to_save = socialiqa.map(
    lambda row, idx: {'text': format_row(row), 'id': idx}, 
    remove_columns=['context', 'question', 'answer'],
    with_indices=True
)
socialiqa_to_save
```

```python
socialiqa_to_save.to_csv("qa/socialiqa/train_qac.csv")
```
